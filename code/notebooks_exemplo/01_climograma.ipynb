## 1. Os dados
A normal climatológica é a média de 30 anos de cada mês. Aqui, 1991-2020.

```python
# Normal climatológica 1991-2020 (base do Xavier), janeiro a dezembro

# ---------- Santa Helena-PR ----------
latitude_sh = -24.86
temperatura_sh = [27.0, 27.0, 26.3, 23.9, 20.3, 19.3, 18.9, 20.8, 22.6, 24.6, 25.4, 26.5]   # °C
chuva_sh = [172.4, 157.2, 131.5, 140.0, 183.1, 140.4, 100.4, 84.1, 139.4, 227.8, 186.9, 186.6]   # mm

# ---------- Curitiba-PR ----------
latitude_ct = -25.43
temperatura_ct = [22.4, 22.5, 21.4, 19.6, 16.4, 15.3, 14.8, 16.0, 17.2, 18.8, 20.0, 21.7]   # °C
chuva_ct = [205.8, 177.0, 142.3, 80.5, 88.0, 109.3, 98.3, 82.5, 136.1, 160.7, 119.4, 152.1]   # mm

meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
```

## 2. Climograma de Santa Helena
Barras = chuva; linha = temperatura.

```python
# Carrega a biblioteca de gráficos com o apelido "plt"
import matplotlib.pyplot as plt

# Cria a figura e o primeiro eixo (o da chuva)
fig, eixo_chuva = plt.subplots(figsize=(8, 4))

# Barras de chuva
eixo_chuva.bar(meses, chuva_sh, color='steelblue')
eixo_chuva.set_ylabel('Chuva (mm)')

# Segundo eixo vertical, à direita, para a temperatura
eixo_temperatura = eixo_chuva.twinx()
eixo_temperatura.plot(meses, temperatura_sh, color='red', marker='o')
eixo_temperatura.set_ylabel('Temperatura (°C)')

eixo_chuva.set_title('Climograma de Santa Helena-PR (1991-2020)')
plt.show()
```

## 3. As duas cidades lado a lado
`plt.subplots(1, 2)` cria **1 linha e 2 colunas** de gráficos.

```python
fig, eixos = plt.subplots(1, 2, figsize=(13, 4))

# ----- Gráfico da esquerda: Santa Helena (eixos[0]) -----
eixos[0].bar(meses, chuva_sh, color='steelblue')
eixos[0].set_ylabel('Chuva (mm)')
eixos[0].set_ylim(0, 250)                        # mesma escala nos dois gráficos
eixo_t0 = eixos[0].twinx()
eixo_t0.plot(meses, temperatura_sh, color='red', marker='o')
eixo_t0.set_ylim(10, 30)
eixos[0].set_title('Santa Helena-PR')

# ----- Gráfico da direita: Curitiba (eixos[1]) -----
eixos[1].bar(meses, chuva_ct, color='steelblue')
eixos[1].set_ylim(0, 250)
eixo_t1 = eixos[1].twinx()
eixo_t1.plot(meses, temperatura_ct, color='red', marker='o')
eixo_t1.set_ylim(10, 30)
eixo_t1.set_ylabel('Temperatura (°C)')
eixos[1].set_title('Curitiba-PR')

plt.tight_layout()
plt.show()
```

> 🧠 **Repare:** usamos a **mesma escala** (`set_ylim`) nos dois gráficos. Sem isso, cada gráfico escolhe a
> sua escala, e a comparação visual engana.
