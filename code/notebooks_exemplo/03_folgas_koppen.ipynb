**Folga = valor do clima − limiar da regra.** Positiva: acima do limiar. Negativa: abaixo. Perto de zero:
**perto de mudar**.

```python
# Normal climatológica 1991-2020 (base do Xavier), janeiro a dezembro

# ---------- Santa Helena-PR ----------
latitude_sh = -24.86
temperatura_sh = [27.0, 27.0, 26.3, 23.9, 20.3, 19.3, 18.9, 20.8, 22.6, 24.6, 25.4, 26.5]   # °C
chuva_sh = [172.4, 157.2, 131.5, 140.0, 183.1, 140.4, 100.4, 84.1, 139.4, 227.8, 186.9, 186.6]   # mm

# ---------- Curitiba-PR ----------
latitude_ct = -25.43
temperatura_ct = [22.4, 22.5, 21.4, 19.6, 16.4, 15.3, 14.8, 16.0, 17.2, 18.8, 20.0, 21.7]   # °C
chuva_ct = [205.8, 177.0, 142.3, 80.5, 88.0, 109.3, 98.3, 82.5, 136.1, 160.7, 119.4, 152.1]   # mm

meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
```

## 1. As folgas das duas cidades

```python
# ---------- Santa Helena ----------
folga_A_sh = min(temperatura_sh) - 18                     # A × C: mês mais frio vs 18 °C
folga_a_sh = max(temperatura_sh) - 22                     # a × b: mês mais quente vs 22 °C
folga_f_sh = min(chuva_sh) - 60                           # Af: mês mais seco vs 60 mm
limiar_moncao_sh = 100 - sum(chuva_sh) / 25               # Am: limiar da monção
folga_m_sh = min(chuva_sh) - limiar_moncao_sh

# ---------- Curitiba ----------
folga_A_ct = min(temperatura_ct) - 18
folga_a_ct = max(temperatura_ct) - 22
folga_f_ct = min(chuva_ct) - 60
limiar_moncao_ct = 100 - sum(chuva_ct) / 25
folga_m_ct = min(chuva_ct) - limiar_moncao_ct

print('                 Santa Helena   Curitiba')
print('Folga A/C (°C):  ', round(folga_A_sh, 2), '       ', round(folga_A_ct, 2))
print('Folga a/b (°C):  ', round(folga_a_sh, 2), '        ', round(folga_a_ct, 2))
print('Folga Af  (mm):  ', round(folga_f_sh, 1), '        ', round(folga_f_ct, 1))
print('Folga Am  (mm):  ', round(folga_m_sh, 1), '        ', round(folga_m_ct, 1))
```

## 2. O gráfico das folgas de temperatura

```python
import matplotlib.pyplot as plt

# Nomes dos critérios (eixo x)
criterios = ['A/C (mês frio)', 'a/b (mês quente)']

# Posição das barras: Santa Helena um pouco à esquerda, Curitiba um pouco à direita
posicoes = [0, 1]
posicoes_sh = [0 - 0.2, 1 - 0.2]
posicoes_ct = [0 + 0.2, 1 + 0.2]

fig, eixo = plt.subplots(figsize=(7, 4))
eixo.bar(posicoes_sh, [folga_A_sh, folga_a_sh], width=0.4, color='tomato', label='Santa Helena')
eixo.bar(posicoes_ct, [folga_A_ct, folga_a_ct], width=0.4, color='steelblue', label='Curitiba')

# Linha no zero = o limiar. Barra perto da linha = perto de mudar.
eixo.axhline(0, color='black')
eixo.set_xticks(posicoes)
eixo.set_xticklabels(criterios)
eixo.set_ylabel('Folga (°C)')
eixo.set_title('Quanto falta para mudar? (1991-2020)')
eixo.legend()
plt.show()
```

## 3. A folga do grupo B (seco)
O limiar de aridez depende da temperatura média anual (T) e de **quando** chove (verão = out a mar).

```python
# Chuva do verão em Santa Helena: out (posição 9), nov (10), dez (11), jan (0), fev (1), mar (2)
chuva_verao_sh = chuva_sh[9] + chuva_sh[10] + chuva_sh[11] + chuva_sh[0] + chuva_sh[1] + chuva_sh[2]
fracao_verao_sh = chuva_verao_sh / sum(chuva_sh)
temperatura_media_sh = sum(temperatura_sh) / 12

# O limiar muda conforme a fração de chuva no verão
if fracao_verao_sh >= 0.7:
    limiar_B_sh = 10 * (2 * temperatura_media_sh + 28)
elif fracao_verao_sh <= 0.3:
    limiar_B_sh = 10 * (2 * temperatura_media_sh)
else:
    limiar_B_sh = 10 * (2 * temperatura_media_sh + 14)

folga_B_sh = sum(chuva_sh) - limiar_B_sh
print('Fração da chuva no verão:', round(fracao_verao_sh, 2))
print('Limiar de aridez (mm):', round(limiar_B_sh))
print('Folga B (mm):', round(folga_B_sh))
```
