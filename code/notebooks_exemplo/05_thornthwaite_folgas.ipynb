```python
# Normal climatológica 1991-2020 (base do Xavier), janeiro a dezembro

# ---------- Santa Helena-PR ----------
latitude_sh = -24.86
temperatura_sh = [27.0, 27.0, 26.3, 23.9, 20.3, 19.3, 18.9, 20.8, 22.6, 24.6, 25.4, 26.5]   # °C
chuva_sh = [172.4, 157.2, 131.5, 140.0, 183.1, 140.4, 100.4, 84.1, 139.4, 227.8, 186.9, 186.6]   # mm

# ---------- Curitiba-PR ----------
latitude_ct = -25.43
temperatura_ct = [22.4, 22.5, 21.4, 19.6, 16.4, 15.3, 14.8, 16.0, 17.2, 18.8, 20.0, 21.7]   # °C
chuva_ct = [205.8, 177.0, 142.3, 80.5, 88.0, 109.3, 98.3, 82.5, 136.1, 160.7, 119.4, 152.1]   # mm

meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
```

## 1. Classificação de Thornthwaite
A classe tem 4 partes: **umidade** (pelo Im), **subtipo** (Ia/Ih), **térmica** (ETP anual) e
**concentração** da ETP no verão.

```python
import agrometeorologiapy as amp
import pandas as pd

# ---------- Santa Helena ----------
# 1) ETP de Thornthwaite
tabela_etp_sh = amp.thornthwaite_mensal(pd.DataFrame({'T_media_C': temperatura_sh}), lat=latitude_sh)
etp_sh = tabela_etp_sh['ETP_mm_mes']

# 2) Classificação (a função faz o balanço hídrico por dentro, CAD = 100 mm)
th_sh = amp.classificacao_thornthwaite(chuva_sh, etp_sh, latitude_sh, CAD=100)

print('Classe:', th_sh['classe'], '-', th_sh['descricao'])
print('Im (%):', round(th_sh['Im'], 1), '| Ih (%):', round(th_sh['Ih'], 1), '| Ia (%):', round(th_sh['Ia'], 1))
print('ETP anual (mm):', round(th_sh['ETP_anual']))
```

```python
# ---------- Curitiba ----------
tabela_etp_ct = amp.thornthwaite_mensal(pd.DataFrame({'T_media_C': temperatura_ct}), lat=latitude_ct)
etp_ct = tabela_etp_ct['ETP_mm_mes']
th_ct = amp.classificacao_thornthwaite(chuva_ct, etp_ct, latitude_ct, CAD=100)

print('Classe:', th_ct['classe'], '-', th_ct['descricao'])
print('Im (%):', round(th_ct['Im'], 1), '| ETP anual (mm):', round(th_ct['ETP_anual']))
```

## 2. Os limites de classe

## 3. As folgas, olhando a tabela
Encontre na tabela os dois limites entre os quais o valor está e digite-os.

```python
# Santa Helena: o Im está entre quais limites? (olhe o valor impresso acima e a tabela)
limite_de_baixo_im_sh = 40
limite_de_cima_im_sh = 60

folga_im_para_baixo_sh = th_sh['Im'] - limite_de_baixo_im_sh
folga_im_para_cima_sh = limite_de_cima_im_sh - th_sh['Im']

print('Santa Helena: o Im pode cair', round(folga_im_para_baixo_sh, 1), 'pontos ou subir',
      round(folga_im_para_cima_sh, 1), 'pontos antes de mudar de classe')
```

> ⚠️ Os limites `40` e `60` acima valem para o Im de Santa Helena impresso na seção 1. **Confira** se ele
> está mesmo entre 40 e 60. Para Curitiba, olhe o valor dela e escolha os limites certos.

```python
# Curitiba: troque pelos limites corretos para o Im de Curitiba
limite_de_baixo_im_ct = 60
limite_de_cima_im_ct = 80

folga_im_para_baixo_ct = th_ct['Im'] - limite_de_baixo_im_ct
folga_im_para_cima_ct = limite_de_cima_im_ct - th_ct['Im']

print('Curitiba: o Im pode cair', round(folga_im_para_baixo_ct, 1), 'pontos ou subir',
      round(folga_im_para_cima_ct, 1), 'pontos antes de mudar de classe')
```
