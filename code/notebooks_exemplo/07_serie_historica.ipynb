```python
# Endereço da pasta de dados do projeto no GitHub (arquivos "raw" = texto puro).
# O pandas consegue ler um CSV direto de um endereço da internet.
URL_DADOS = 'https://raw.githubusercontent.com/fcoliveira-utfpr/change_climate/main/code/dados'
```

## 1. Ler e olhar os dados

```python
import pandas as pd

# Lê as séries mensais (1961-2025) direto do GitHub.
#   parse_dates=['data'] -> a coluna 'data' é entendida como data
#   index_col='data'     -> a coluna 'data' vira o índice (o "nome" de cada linha)
serie_sh = pd.read_csv(f'{URL_DADOS}/santa_helena_mensal.csv', parse_dates=['data'], index_col='data')
serie_ct = pd.read_csv(f'{URL_DADOS}/curitiba_mensal.csv', parse_dates=['data'], index_col='data')

LATITUDE_SH = -24.86
LATITUDE_CT = -25.43

serie_sh.head()
```

```python
# Sempre olhe os dados antes de usar
print('Número de meses:', len(serie_sh))
print('Primeiro mês:', serie_sh.index[0].date(), '| último mês:', serie_sh.index[-1].date())
serie_sh.describe().round(1)
```

## 2. Da série mensal para a série anual

```python
# Temperatura média de cada ano = média dos 12 meses; chuva de cada ano = SOMA dos 12 meses
#   resample('YS') agrupa os meses por ano ("Year Start")
temperatura_anual_sh = serie_sh['tmed'].resample('YS').mean()
chuva_anual_sh = serie_sh['chuva'].resample('YS').sum()

temperatura_anual_ct = serie_ct['tmed'].resample('YS').mean()
chuva_anual_ct = serie_ct['chuva'].resample('YS').sum()

temperatura_anual_sh.head()
```

```python
import matplotlib.pyplot as plt

fig, eixos = plt.subplots(2, 1, figsize=(11, 7), sharex=True)

# .index.year pega só o ano de cada data
eixos[0].plot(temperatura_anual_sh.index.year, temperatura_anual_sh, marker='o', color='tomato', label='Santa Helena')
eixos[0].plot(temperatura_anual_ct.index.year, temperatura_anual_ct, marker='o', color='steelblue', label='Curitiba')
eixos[0].set_ylabel('Temperatura média anual (°C)')
eixos[0].legend()
eixos[0].grid(alpha=0.3)

eixos[1].plot(chuva_anual_sh.index.year, chuva_anual_sh, marker='o', color='tomato', label='Santa Helena')
eixos[1].plot(chuva_anual_ct.index.year, chuva_anual_ct, marker='o', color='steelblue', label='Curitiba')
eixos[1].set_ylabel('Chuva anual (mm)')
eixos[1].set_xlabel('Ano')
eixos[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

## 3. Primeiras perguntas

```python
print('Santa Helena, ano mais quente:', temperatura_anual_sh.idxmax().year, '-', round(temperatura_anual_sh.max(), 2), '°C')
print('Curitiba, ano mais quente:', temperatura_anual_ct.idxmax().year, '-', round(temperatura_anual_ct.max(), 2), '°C')

# Média de duas décadas, para comparar
media_1960_sh = temperatura_anual_sh.loc['1961':'1970'].mean()
media_2010_sh = temperatura_anual_sh.loc['2016':'2025'].mean()
print('Santa Helena: 1961-1970 =', round(media_1960_sh, 2), '°C | 2016-2025 =', round(media_2010_sh, 2), '°C')
```
