```python
# Normal climatológica 1991-2020 (base do Xavier), janeiro a dezembro

# ---------- Santa Helena-PR ----------
latitude_sh = -24.86
temperatura_sh = [27.0, 27.0, 26.3, 23.9, 20.3, 19.3, 18.9, 20.8, 22.6, 24.6, 25.4, 26.5]   # °C
chuva_sh = [172.4, 157.2, 131.5, 140.0, 183.1, 140.4, 100.4, 84.1, 139.4, 227.8, 186.9, 186.6]   # mm

# ---------- Curitiba-PR ----------
latitude_ct = -25.43
temperatura_ct = [22.4, 22.5, 21.4, 19.6, 16.4, 15.3, 14.8, 16.0, 17.2, 18.8, 20.0, 21.7]   # °C
chuva_ct = [205.8, 177.0, 142.3, 80.5, 88.0, 109.3, 98.3, 82.5, 136.1, 160.7, 119.4, 152.1]   # mm

meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
```

## 1. Os números do clima (Santa Helena)

```python
# Chuva anual: soma dos 12 meses
chuva_anual_sh = sum(chuva_sh)

# Temperatura média anual: soma dividida por 12
temperatura_media_sh = sum(temperatura_sh) / 12

# Mês mais frio, mês mais quente e mês mais seco
t_mes_frio_sh = min(temperatura_sh)
t_mes_quente_sh = max(temperatura_sh)
p_mes_seco_sh = min(chuva_sh)

print('Chuva anual (mm):', round(chuva_anual_sh, 1))
print('Temperatura média anual (°C):', round(temperatura_media_sh, 1))
print('Mês mais frio (°C):', t_mes_frio_sh)
print('Mês mais quente (°C):', t_mes_quente_sh)
print('Mês mais seco (mm):', p_mes_seco_sh)
```

## 2. Köppen à mão (com `if`)
Antes da biblioteca, as duas primeiras regras escritas por nós (o grupo B fica para depois):
- mês mais frio **≥ 18 °C** → **A** (tropical); senão → **C** (temperado);
- mês mais quente **≥ 22 °C** → 3ª letra **a**; senão → **b**.

```python
# "if" quer dizer "se", "else" quer dizer "senão".
# As linhas recuadas (4 espaços) pertencem ao if ou ao else.
if t_mes_frio_sh >= 18:
    print('Santa Helena: grupo A (tropical)')
else:
    print('Santa Helena: grupo C (temperado)')

if t_mes_quente_sh >= 22:
    print('Santa Helena: verão quente (a)')
else:
    print('Santa Helena: verão ameno (b)')
```

## 3. Köppen com a biblioteca

```python
import agrometeorologiapy as amp

# A função recebe as 12 temperaturas, as 12 chuvas e a latitude
resultado_sh = amp.classificacao_koppen(temperatura_sh, chuva_sh, latitude_sh)
resultado_ct = amp.classificacao_koppen(temperatura_ct, chuva_ct, latitude_ct)

# O resultado é um dicionário: pegamos o valor guardado com o nome 'classe'
print('Santa Helena:', resultado_sh['classe'])
print('Curitiba:', resultado_ct['classe'])
```

## 4. Experimento: o que faz a classe mudar?

```python
# Copiamos a lista para não estragar a original. list() faz uma cópia.
temperatura_teste = list(temperatura_ct)

# Diminui 1 °C em fevereiro (posição 1; as posições começam em 0: jan = 0, fev = 1, ...)
temperatura_teste[1] = temperatura_teste[1] - 1

print('Fevereiro original:', temperatura_ct[1], '| teste:', temperatura_teste[1])
print('Classe com o teste:', amp.classificacao_koppen(temperatura_teste, chuva_ct, latitude_ct)['classe'])
```
